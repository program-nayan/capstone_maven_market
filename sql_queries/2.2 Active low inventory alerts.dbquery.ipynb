```sql
SELECT 
    store_id,
    product_id,
    stock_on_hand,
    reorder_level,
    restock_flag,
    last_updated
FROM maven_market_uc.silver.silver_inventory
QUALIFY ROW_NUMBER() OVER (
    PARTITION BY store_id, product_id 
    ORDER BY last_updated DESC, _ingested_at DESC
) = 1
WHERE restock_flag = TRUE OR stock_on_hand <= reorder_level
ORDER BY stock_on_hand ASC;
```
