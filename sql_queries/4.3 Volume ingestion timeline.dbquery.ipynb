```sql
SELECT 
    DATE(timestamp) AS execution_date,
    pipeline_name,
    SUM(records_processed) AS daily_records_processed,
    ROUND(AVG(execution_time_seconds), 4) AS avg_execution_time_sec
FROM maven_market_uc.audit.audit_logs
WHERE status = 'SUCCESS'
GROUP BY 1, 2
ORDER BY execution_date ASC;
```
