```sql
SELECT 
    pipeline_name,
    step_name,
    SUM(records_processed) AS total_valid_records,
    MAX(timestamp) AS last_processed_at,
    get_json_object(metadata, '$.expectations_summary') AS quality_expectations
FROM maven_market_uc.audit.audit_logs
WHERE status = 'SUCCESS' 
  AND pipeline_name LIKE 'dlt_pipeline%'
GROUP BY pipeline_name, step_name, metadata
ORDER BY last_processed_at DESC;
```
