```sql
-- Operational Record Ingestion Rates
SELECT 
    DATE_TRUNC('hour', timestamp) AS execution_hour,
    pipeline_name,
    SUM(records_processed) AS records_ingested,
    SUM(CASE WHEN status = 'FAILED' THEN 1 ELSE 0 END) AS failed_job_count
FROM maven_market_uc.audit.audit_logs
WHERE timestamp >= NOW() - INTERVAL 7 DAYS
GROUP BY 1, 2
ORDER BY execution_hour DESC;
```
