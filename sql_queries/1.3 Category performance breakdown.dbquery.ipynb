```sql
SELECT 
    COALESCE(p.product_brand, 'Unbranded') AS product_brand,
    ROUND(SUM(s.daily_revenue), 2) AS brand_revenue,
    SUM(s.total_units_sold) AS total_units_sold
FROM maven_market_uc.gold.fact_sales s
JOIN maven_market_uc.gold.dim_products p 
    ON s.product_id = p.product_id
GROUP BY 1
ORDER BY brand_revenue DESC;
```
