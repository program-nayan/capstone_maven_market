```sql
CREATE SCHEMA IF NOT EXISTS maven_market_uc.audit;

CREATE TABLE IF NOT EXISTS maven_market_uc.audit.audit_logs (
    timestamp TIMESTAMP NOT NULL COMMENT 'UTC event timestamp',
    environment STRING NOT NULL COMMENT 'Environment: dev, staging, prod',
    pipeline_name STRING NOT NULL COMMENT 'Pipeline or job identifier',
    step_name STRING NOT NULL COMMENT 'Granular step or table module',
    status STRING NOT NULL COMMENT 'Status: STARTED, SUCCESS, FAILED',
    records_processed BIGINT NOT NULL COMMENT 'Row count mutated or transformed',
    error_message STRING COMMENT 'Error stacktrace if failed',
    metadata STRING COMMENT 'JSON metadata payload'
)
USING DELTA
PARTITIONED BY (environment, status);
```
