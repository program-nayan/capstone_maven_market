```sql
SELECT 
    ROUND(SUM(s.daily_revenue), 2) AS total_revenue,
    ROUND(
        SUM(s.daily_revenue - (s.total_units_sold * COALESCE(p.product_cost, p.product_retail_price * 0.6))), 
        2
    ) AS total_profit,
    ROUND(
        (SUM(s.daily_revenue - (s.total_units_sold * COALESCE(p.product_cost, p.product_retail_price * 0.6))) / 
         NULLIF(SUM(s.daily_revenue), 0)) * 100, 
        2
    ) AS profit_margin_pct
FROM maven_market_uc.gold.fact_sales s
LEFT JOIN maven_market_uc.gold.dim_products p 
    ON s.product_id = p.product_id;
```
