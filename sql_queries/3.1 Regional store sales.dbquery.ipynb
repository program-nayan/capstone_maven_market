```sql
SELECT 
    COALESCE(r.sales_region, st.store_state) AS store_region,
    st.store_name,
    st.store_city,
    st.store_state,
    ROUND(SUM(s.daily_revenue), 2) AS regional_revenue,
    SUM(s.order_count) AS total_transactions
FROM maven_market_uc.gold.fact_sales s
JOIN maven_market_uc.gold.dim_stores st 
    ON s.store_id = st.store_id
LEFT JOIN maven_market_uc.bronze.regions r
    ON st.region_id = r.region_id
GROUP BY 1, 2, 3, 4
ORDER BY regional_revenue DESC;
```
