```sql
SELECT 
    st.store_name,
    st.store_city,
    ROUND(SUM(s.daily_revenue), 2) AS store_revenue,
    RANK() OVER (ORDER BY SUM(s.daily_revenue) DESC) AS regional_rank
FROM maven_market_uc.gold.fact_sales s
JOIN maven_market_uc.gold.dim_stores st 
    ON s.store_id = st.store_id
GROUP BY 1, 2
LIMIT 10;
```
