```sql
SELECT 
    step_name AS stream_name,
    SUM(records_processed) AS total_records_ingested,
    ROUND(AVG(execution_time_seconds), 4) AS avg_duration_sec,
    MAX(timestamp) AS last_stream_sync_utc
FROM maven_market_uc.audit.audit_logs
WHERE timestamp >= CURRENT_TIMESTAMP() - INTERVAL 24 HOURS
GROUP BY step_name;
```
