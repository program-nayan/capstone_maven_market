```sql
-- Pipeline Health & Execution Metrics (Last 24 Hours)
SELECT 
    pipeline_name,
    step_name,
    status,
    COUNT(*) AS total_runs,
    SUM(records_processed) AS total_records_processed,
    MAX(timestamp) AS last_execution_time,
    COLLECT_SET(error_message) AS recent_errors
FROM maven_market_uc.audit.audit_logs
WHERE timestamp >= NOW() - INTERVAL 24 HOURS
GROUP BY pipeline_name, step_name, status
ORDER BY last_execution_time DESC;
```
