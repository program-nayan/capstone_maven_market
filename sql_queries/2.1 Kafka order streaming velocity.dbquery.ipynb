```sql
SELECT 
    DATE_TRUNC('minute', order_date) AS streaming_minute,
    COUNT(DISTINCT order_id) AS orders_per_minute,
    SUM(quantity) AS units_per_minute
FROM maven_market_uc.silver.silver_orders
WHERE order_date >= CURRENT_TIMESTAMP() - INTERVAL 24 HOURS
GROUP BY 1
ORDER BY streaming_minute ASC;
```
